```python
P = [[0 for _ in range(10)] for _ in range(8)]

for i, N in enumerate(range(3, 10)):
    for j, beta in enumerate(np.linspace(0.5, 4.5, 10)):
        beta /= N
        gauge_model = GaugeLattice(grp = ZN(N), M = 4, beta = beta, start = 'cold')
        gauge_model.sim(100)
        hist, _ = gauge_model.sim(100)
        plt.plot(hist, label = f"N={N}, beta={beta:.2f}")
        P[i][j] = np.mean(hist)

plt.legend()
plt.show()
```

```python
P = np.array(P)
for i, N in enumerate(range(3, 10)):
    plt.plot(np.linspace(0.5, 4.5, 10), P[i], label = f"N={N}")
plt.legend()
```
